```python
import json, os, subprocess, torch
out={'cuda_visible_devices': os.getenv('CUDA_VISIBLE_DEVICES'), 'torch_version': torch.__version__, 'torch_cuda_version': torch.version.cuda, 'torch_cuda_available': bool(torch.cuda.is_available())}
try:
    p=subprocess.run(['nvidia-smi','--query-gpu=name,compute_cap','--format=csv,noheader'], text=True, capture_output=True, timeout=20)
    out['nvidia_smi_rc']=p.returncode; out['nvidia_smi']=p.stdout.strip(); out['nvidia_smi_err']=p.stderr.strip()
except Exception as e:
    out['nvidia_smi_error']=repr(e)
if torch.cuda.is_available():
    out['device_name']=torch.cuda.get_device_name(0); out['device_capability']=list(torch.cuda.get_device_capability(0))
    try:
        x=torch.tensor([1.0],device='cuda'); out['cuda_op_ok']=float((x+1).item())==2.0
    except Exception as e:
        out['cuda_op_ok']=False; out['cuda_op_error']=repr(e)
open('/kaggle/working/gpu-probe-notebook.json','w').write(json.dumps(out,sort_keys=True,indent=2)+'\n')
print('NEXUS_NOTEBOOK_GPU_PROBE='+json.dumps(out,sort_keys=True))
```
